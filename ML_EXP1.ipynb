```python
# Import required libraries
# Task 1: Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Upload train.csv in Google Colab

from google.colab import files

uploaded = files.upload()  # Select train.csv from your computer

# Load the uploaded CSV file into a Pandas DataFrame
df = pd.read_csv("Titanic-Dataset.csv")

# Display the first 5 rows
df.head()
```

```python
# Task 3: Display first 10 rows
print("First 10 Rows:")
print(df.head(10))

# Task 4: Display last 5 rows
print("\nLast 5 Rows:")
print(df.tail(5))

# Task 5: Check dataset dimensions
print("\nDataset Shape:")
print(df.shape)   # Expected: (891, 12)

# Task 6: Display dataset information
print("\nDataset Information:")
df.info()

# Task 7: Generate summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Display mean, minimum, and maximum values of Age and Fare
print("\nAge Statistics:")
print("Mean Age:", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\nFare Statistics:")
print("Mean Fare:", df["Fare"].mean())
print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())

# Task 8: Identify missing values
print("\nMissing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Task 9: Fill missing Age values with the median
df["Age"].fillna(df["Age"].median(), inplace=True)

# Check if missing values in Age are filled
print("Missing values in Age after filling:")
print(df["Age"].isnull().sum())

# Task 10: Filter surviving passengers (Survived == 1)
survivors = df[df["Survived"] == 1]

print("\nSurviving Passengers:")
print(survivors.head())

# Task 11: Convert Fare column to a NumPy array
fare_array = df["Fare"].to_numpy()

print("\nFare as NumPy Array:")
print(fare_array)

# Task 12: Calculate maximum and average Fare
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print("\nMaximum Fare:", max_fare)
print("Average Fare:", avg_fare)
```

```python
# Task 13: Plot histogram of Age

plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20, edgecolor="black")
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.grid(True)
plt.show()


# Task 14: Plot bar chart of Survival Count

plt.figure(figsize=(6,4))
df["Survived"].value_counts().plot(kind="bar")
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=0)
plt.show()


# Task 15: Plot scatter graph of Age vs Fare

plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Fare"])
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.grid(True)
plt.show()
```
